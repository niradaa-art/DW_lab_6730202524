หาจำนวนผู้ป่วยที่เป็นคน "คนต่างชาติเดินทางมาจากต่างประเทศ

````
this_data = data_covid[['sex','age','province_of_onset']]
print(this_data)
````

````
female = this_data[(this_data['sex']=='หญิง')]
````

````
female_age_numeric = pd.to_numeric(female['age'], errors='coerce')
mean_age = female_age_numeric.dropna().mean()

print(f'อายุเฉลี่ย ของ ผู้ป่วยหญิง {mean_age}')
````

````
male = this_data[(this_data['sex']=='ชาย')&(this_data['sex']=='ชาย')]
````

````
male_age_numeric = pd.to_numeric(male['age'], errors='coerce')
mean_age = male_age_numeric.dropna().mean()

print(f'อายุเฉลี่ย ของ ผู้ป่วยชาย {mean_age}')
````

````
female_KK = female[female['province_of_onset']=='ขอนแก่น']
female_KK_age_numeric = pd.to_numeric(female_KK['age'], errors='coerce')

sum_age_kk = 0
N_age_kk = 0
for a in female_KK_age_numeric:
    if a > 0:
        sum_age_kk += a
        N_age_kk += 1

if N_age_kk > 0:
    print(f'อายุเฉลี่ย ของ ผู้ป่วยหญิง ในจังหวัดขอนแก่น {sum_age_kk/N_age_kk}')
else:
    print('ไม่มีข้อมูล')
````

````
male_KK = male[male['province_of_onset']=='ขอนแก่น']
male_KK_age_numeric = pd.to_numeric(male_KK['age'], errors='coerce')

sum_age_kk = 0
N_age_kk = 0
for a in male_KK_age_numeric:
    if a > 0:
        sum_age_kk += a
        N_age_kk += 1

if N_age_kk > 0:
    print(f'อายุเฉลี่ย ของ ผู้ป่วยชาย ในจังหวัดขอนแก่น {sum_age_kk/N_age_kk}')
else:
    print('ไม่มีข้อมูล')
````
